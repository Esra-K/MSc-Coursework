# Prepare data

download data from https://forexsb.com/historical-forex-data

## read dataset

```python
import pandas as pd

df = pd.read_csv('USDCAD_D1.csv', sep='\t')
df.columns = [c.lower() for c in df.columns]
df.head()
```

## data visualization

```python
import plotly.graph_objects as go

start, end = -200, None
fig = go.Figure(data=[
    go.Candlestick(
      x=df.iloc[start:end]['time'],
      open=df.iloc[start:end]['open'],
      high=df.iloc[start:end]['high'],
      low=df.iloc[start:end]['low'],
      close=df.iloc[start:end]['close']
    )
])

fig.show()
```

# Methodologies

## Mathod 1: Employ Prompt engineering


This method which is our baseline method, first converts the signal parts into a natural language sentences and then fine-tunes an LLM on them for the next sentence prediction.

## Method 2: Think of signals as a language

This method quantizes the signal and creates a vocabulary to be used as tokens of the language of the signal. Then LLM architecture will be trained for the next token prediction.
